# Data Loading

```python
import numpy as np
import scipy
import os

print('SciPy', scipy.__version__)
print('NumPy', np.__version__)
```

```python
DATASET_PATH = os.path.join("..", "..", "Dataset")
RAW_DATA_PATH = os.path.join(DATASET_PATH, "NovAtel", "GPS_data_combined_Nova.mat")
REF_POS_DATA_PATH = os.path.join(DATASET_PATH, "NovAtel", "2.BESTGPSPOS.mat")
REF_VEL_DATA_PATH = os.path.join(DATASET_PATH, "NovAtel", "2.BESTGPSVEL.mat")
```

```python
from scipy.io import loadmat
```

```python
raw_data_mat = loadmat(RAW_DATA_PATH, struct_as_record=False, squeeze_me=True)
print(raw_data_mat.keys())
```

```python
ref_pos_data_mat = loadmat(REF_POS_DATA_PATH, struct_as_record=False, squeeze_me=True)
print(ref_pos_data_mat.keys())
```

```python
ref_vel_data_mat = loadmat(REF_VEL_DATA_PATH, struct_as_record=False, squeeze_me=True)
print(ref_vel_data_mat.keys())
```

```python
satellite_data = {
    'raw_data': {
        'time_grid': raw_data_mat['GPS_sec'],
        'sat_x': np.array(raw_data_mat['sat_x'], dtype=object),
        'sat_y': np.array(raw_data_mat['sat_y'], dtype=object),
        'sat_z': np.array(raw_data_mat['sat_z'], dtype=object),
        'sat_vx': np.array(raw_data_mat['sat_vx'], dtype=object),
        'sat_vy': np.array(raw_data_mat['sat_vy'], dtype=object),
        'sat_vz': np.array(raw_data_mat['sat_vz'], dtype=object),
        'rho': np.array(raw_data_mat['rho_C1'], dtype=object),
        'rho_std': np.array(raw_data_mat['rho_C1_std'], dtype=object),
        'dt_sv': np.array(raw_data_mat['dt_sv'], dtype=object),
        'doppler': np.array(raw_data_mat['dopp_L1'], dtype=object),
        'az_sat': np.array(raw_data_mat['az_sat'], dtype=object),
        'el_sat': np.array(raw_data_mat['el_sat'], dtype=object),
    },
    'reference_data': {
        'ref_alt': ref_pos_data_mat['BGP_Alt'],
        'ref_alt_std': ref_pos_data_mat['BGP_Alt_std'],
        'ref_lat': ref_pos_data_mat['BGP_Lat'],
        'ref_lat_std': ref_pos_data_mat['BGP_Lat_std'],
        'ref_long': ref_pos_data_mat['BGP_Long'],
        'ref_long_std': ref_pos_data_mat['BGP_Long_std'],
        'ref_ve': ref_vel_data_mat['BGV_ve'],
        'ref_vn': ref_vel_data_mat['BGV_vn'],
        'ref_vu': ref_vel_data_mat['BGV_vu'],
    }
}
```

```python
np.savez_compressed('satellite_data.npz', **satellite_data)
```
