## Calibration - Bias Estimates

```python
init_pitch = ref_data['ref_pitch'][0,0]
init_roll = ref_data['ref_roll'][0,0]
init_lat = ref_data['ref_lat'][0,0]
init_alt = ref_data['ref_alt'][0,0]

omega_e = 7.292115e-5 # Earth’s rotation rate (rad/s)
```

```python
def normal_gravity(latitude, height):
    # Constants for the GRS80 (1980 Geodetic Reference System)
    a1 = 9.7803267714   # Gravity at the equator (m/s^2)
    a2 = 0.0052790414   # Coefficient for sin^2(latitude)
    a3 = 0.0000232718   # Coefficient for sin^4(latitude)
    a4 = -0.0000030876910891  # Coefficient for height correction (h)
    a5 = 0.0000000043977311  # Coefficient for height correction (sin^2(latitude) * h)
    a6 = 0.0000000000007211  # Coefficient for height squared (h^2)

    # Calculate normal gravity using the Somigliana-Pizzetti formula
    gamma = (
        a1 * (1 + a2 * np.sin(latitude)**2 + a3 * np.sin(latitude)**4)
        + (a4 + a5 * np.sin(latitude)**2) * height
        + a6 * height**2
    )

    return gamma

g = normal_gravity(init_lat, init_alt)
```

```python
def find_indices(timestamps, start_time, end_time):
    start_idx = np.searchsorted(timestamps, start_time, side='left') # Find first >= start_time
    end_idx = np.searchsorted(timestamps, end_time, side='right') -1  # Find last <= end_time
    return start_idx, end_idx
```

## Sensor Data Resampling

```python
def downsample_and_append(data, sample_rate):
    """
    Downsamples a 2D array with one singleton dimension and one non-singleton dimension by averaging.
    Truncates data for downsampling, appends the truncated portion to the result.

    Parameters:
        data (np.ndarray): Input 2D array with shape (non_singleton, 1) or (1, non_singleton).
        sample_rate (int): The rate at which to downsample the data.

    Returns:
        np.ndarray: The downsampled data with the truncated portion appended.
    """
    # Ensure input has one non-singleton dimension
    if data.shape[0] == 1:
        data = data.flatten()  # Flatten to a 1D array
        length = data.shape[0]  # Non-singleton dimension length
    elif data.shape[1] == 1:
        data = data.flatten()  # Flatten to a 1D array
        length = data.shape[0]  # Non-singleton dimension length
    else:
        raise ValueError("Input array must have one singleton dimension and one non-singleton dimension.")

    # Calculate truncation length
    truncate_length = (length // sample_rate) * sample_rate

    # Truncate the data for downsampling
    truncated_data = data[:truncate_length]

    # Reshape into chunks of size `sample_rate` and compute the mean
    reshaped_data = truncated_data.reshape(-1, sample_rate)
    downsampled_data = np.mean(reshaped_data, axis=1)

    # Get the portion of data that was truncated
    remainder_data = data[truncate_length:]

    # Combine downsampled data and remainder
    result = np.concatenate([downsampled_data, remainder_data])

    return result
```

```python
# NovAtel Data

IMU1_W_trunc_ds = downsample_and_append(IMU1_w_truncated, 200)
print(IMU1_W_trunc_ds.shape)

N = int(np.ceil(IMU1_f_truncated.shape[0]/200))
IMU1_f_trunc_ds = np.zeros((N, 3))

for i in range(IMU1_f_truncated.shape[1]):
    IMU1_f_trunc_ds[:,i] = downsample_and_append(IMU1_f_truncated[:,i].reshape(-1,1), 200)

print(IMU1_f_trunc_ds.shape)
```
